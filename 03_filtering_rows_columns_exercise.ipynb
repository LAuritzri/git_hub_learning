```python
#6. Select all those rows where the `pH` is greater than 3.0 and less than 3.5. 
df_wine_quality_ph_range = df_wine_quality[(df_wine_quality['pH'] > 3.0) & (df_wine_quality['pH'] < 3.5)] 
df_wine_quality_ph_range
```

```python
#7. Further filter the results from 6 to grab only those rows that have a `residual sugar` less than 2.0.
#Tip: Use backticks (``) to mask column names with spaces in query string.
df_wine_quality_ph_range_low_sugar = df_wine_quality[(df_wine_quality['pH'] > 3.0) & (df_wine_quality['pH'] < 3.5) & (df_wine_quality['residual sugar'] < 2.0)] 
df_wine_quality_ph_range_low_sugar
```
